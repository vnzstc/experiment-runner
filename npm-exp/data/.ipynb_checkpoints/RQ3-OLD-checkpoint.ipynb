```python
import pandas as pd
import numpy as np
import os, re
import statsmodels.api as sm
import statsmodels.formula.api as smf
import itertools
from IPython.display import display
from pathlib import Path
from pathlib import PosixPath
import matplotlib.pyplot as plt
```

```python
EXTENSIONS = {
    'npm': '.tgz',
    'xz': '.tar.xz',
    'brotli': '.tar.br',
    'zstd': '.tar.zst',
    'zopfli': '.tgz'
}
results = '../results/npm-exp/run_table.csv'
samples = '../pkgs/samples'
```

```python
# load results
data = pd.read_csv(results)
data.head()
```

```python
# Preprocessing to split run and rep id; Step not needed in the next version of Experiment Runner 
run_ids = data['__run_id'].to_list()
def extract_run_reps(run_ids):
    run_reps = []
 
    for rid in run_ids:
        run, reps = re.findall(r'\d+', rid)
        run_reps.append({'__run_id' : rid, 'run' : run, 'reps' : reps})
 
    return run_reps

run_reps = extract_run_reps(run_ids)
df_run_reps = pd.DataFrame(run_reps)
data = pd.merge(data, df_run_reps, on=['__run_id'])

data['total-energy'] = data['energy-pkg'] + data['energy-ram']
data.head()
```

```python
# sizes of compressed files 
sizes = {
    key : pd.read_csv(f'{samples}/{key}.csv').rename(columns={'name': 'subject'}) for key in EXTENSIONS.keys()
}
```

```python
import statsmodels.api as smd
import statsmodels.formula.api as smf

# regression as energy = fixed_cost + (energy_per_byte * size)
def fit_regression(algo_name, agg="mean"):
    sub = data[data["alg"] == algo_name]
    grouped = (
        sub.groupby(["subject", "run"])["energy-pkg"]
        .agg(["mean", "std", "count"])
        .reset_index()
    )
    
    complete_data = pd.merge(grouped, sizes[algo_name], on=['subject'])
    y_col = "mean" if agg == "mean" else agg
 
    X = sm.add_constant(complete_data["size"])
    y = complete_data[y_col]
    
    model = sm.OLS(y, X).fit()
    robust = model.get_robustcov_results(cov_type="HC3")
    
    print(f"--- {algo_name} (aggregated, n={len(grouped)} packages) ---")
    print(f"fixed_cost      = {robust.params[0]:.6f}  (SE={robust.bse[0]:.6f})")
    print(f"energy_per_byte = {robust.params[1]:.9f}  (SE={robust.bse[1]:.9f})")
 
    return model, robust, grouped, y
```

```python
model, robust, grouped, y = fit_regression('npm')

fitted = model.fittedvalues
plt.scatter(y, fitted, alpha=0.5)

plt.plot([y.min(), y.max()], [y.min(), y.max()], 'r--')  # 45° line

plt.xlabel("Actual y")
plt.ylabel("Fitted y")
plt.title("Fitted vs Actual")
plt.show()
```
